```python
import statistics
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]
```

```python
mean_age = statistics.mean(ages)
median_age = statistics.median(ages)
```

```python
print("a)Length =",len(ages))
print("b)Total =",sum(ages))
print("c)Mean =",mean_age)
print("d)Minimum =",min(ages))
print("  Minimum =",max(ages))
print("e)Median =",median_age)
print("When mean and median disagree, it means data values are unevenly distributed.")
```

```python
df = pd.read_excel("C:/Users/HP/OneDrive/Desktop/JUPYTER/50_startups_messy.xlsx")
print(df.shape) # size
print(df.isnull().sum()) # completeness
print(df.duplicated().sum()) # uniqueness
```

```python
df = df.drop_duplicates()
df["Marketing Spend"] = df["Marketing Spend"].str.strip().str.title()
df["Marketing Spend" ] = pd.to_numeric(df["Marketing Spend"], 
errors="coerce")
```

```python
random_sample = df.sample(20, random_state=7)

stratified_sample = df.groupby("State").sample(frac=0.2, random_state=7)

print(random_sample)
```

```python
print("Population:")
print(df[["R&D Spend", "Marketing Spend"]].mean())

print("\nRandom Sample:")
print(random_sample[["R&D Spend", "Marketing Spend"]].mean())

print("\nStratified Sample:")
print(stratified_sample[["R&D Spend", "Marketing Spend"]].mean())
```

```python
print("Random sample difference:")
print((random_sample[["R&D Spend", "Marketing Spend"]].mean()
       - df[["R&D Spend", "Marketing Spend"]].mean()).abs())

print("Stratified sample difference:")
print((stratified_sample[["R&D Spend", "Marketing Spend"]].mean()
       - df[["R&D Spend", "Marketing Spend"]].mean()).abs())
```


Marketing Spend is closer to the population mean, while R&D Spend is farther from the population mean
